3. Write a program in which use all the operators we can use in Python

```python
# 1. Arithmetic Operators (Math)
a = 10
b = 3

print("Addition:", a + b)        # 13
print("Subtraction:", a - b)     # 7
print("Multiplication:", a * b)  # 30
print("Division:", a / b)        # 3.33
print("Floor Division:", a // b) # 3
print("Modulus (Remainder):", a % b) # 1
print("Power:", a ** b)          # 1000

# 2. Comparison Operators (True or False answers)
print("Is equal to:", a == b)      # False
print("Not equal to:", a != b)     # True
print("Greater than:", a > b)      # True
print("Less than:", a < b)         # False
print("Greater or equal:", a >= b) # True
print("Less or equal:", a <= b)    # False

# 3. Logical Operators (Combine conditions)
x = True
y = False

print("AND:", x and y) # False (both must be true)
print("OR:", x or y)   # True (at least one is true)
print("NOT:", not x)   # False (reverses the boolean)

# 4. Assignment Operators (Store & update values)
num = 5
print("Initial value:", num)

num += 2  # same as num = num + 2
print("After += 2:", num)  # 7

num -= 1  # same as num = num - 1
print("After -= 1:", num)  # 6

num *= 3  # same as num = num * 3
print("After *= 3:", num)  # 18

num /= 2  # same as num = num / 2
print("After /= 2:", num)  # 9.0
```
